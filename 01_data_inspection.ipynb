# 01 - Basic Dataset Inspection
Raw look at the laptop dataset. **No preprocessing, cleaning, encoding, or modelling here.**

```python
# Import pandas for loading and inspecting tabular data
import pandas as pd

# Show all columns when displaying DataFrames
pd.set_option('display.max_columns', None)
```

## Load the dataset

```python
# Load the raw CSV exactly as it is (no index_col, so any saved index shows up as a column)
df = pd.read_csv('laptop_data.csv')
```

## 1. Dataset dimensions

```python
# Shape = (rows, columns)
print('df.shape:', df.shape)
print('Number of rows:', df.shape[0])
print('Number of columns:', df.shape[1])
```

## 2. Column names

```python
# List every column in order
for i, col in enumerate(df.columns):
    print(i, repr(col))
```

```python
# Check for an index-like column created when a DataFrame index was saved to CSV
index_like = [c for c in df.columns if c.startswith('Unnamed') or c.strip() == '']
print('Index-like columns:', index_like)

# If present, check whether it simply equals 0..n-1 (i.e. a redundant copy of the row index)
for c in index_like:
    print(c, 'equals 0..n-1:', (df[c] == pd.RangeIndex(len(df))).all())
```

## 3. Missing values

```python
# Count null values in every column
print(df.isnull().sum())
```

```python
# Total missing values across the whole dataset
total_missing = df.isnull().sum().sum()
print('Total missing values:', total_missing)
print('Dataset contains missing values:', total_missing > 0)
```

## 4. Data types

```python
# pandas dtype of each column
print(df.dtypes)
```

```python
# Separate numerical columns from string/object columns
numerical_cols = df.select_dtypes(include='number').columns.tolist()
object_cols = df.select_dtypes(include=['object', 'string']).columns.tolist()
print('Numerical columns:', numerical_cols)
print('Object/string columns:', object_cols)
```

## 5. First few rows

```python
# First 5 rows to verify the data loaded correctly
df.head()
```

## 6. General dataset information

```python
# Column count, non-null counts, dtypes and memory usage
df.info()
```

## 7. Duplicate rows

```python
# Number of fully duplicated rows (not removed)
print('Duplicated rows:', df.duplicated().sum())
```

```python
# Duplicates ignoring the index-like column (its unique values would hide real duplicates)
print('Duplicated rows (ignoring index-like columns):', df.drop(columns=index_like).duplicated().sum())
```

## 8. Basic target inspection (`Price`)

```python
# Datatype of the target
print('Price dtype:', df['Price'].dtype)

# Summary statistics of the target (no transformation)
print('Min:   ', df['Price'].min())
print('Max:   ', df['Price'].max())
print('Mean:  ', df['Price'].mean())
print('Median:', df['Price'].median())
print('Std:   ', df['Price'].std())
```
